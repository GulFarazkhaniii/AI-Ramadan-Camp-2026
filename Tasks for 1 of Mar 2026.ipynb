```python
# 1. Make a calculator using Python with addition, subtraction, multiplication, division and power.

Num_1 = int(input("Enter your first number"))
Num_2 = int(input("Enter your second number"))
Operand = input("Enter one of the operator  -, +, Multiply, Divide or ** ")
if Operand == "+":
    print("Sum of ",Num_1," &",Num_2," is", Num_1+Num_2)
elif Operand == "-":
    print("Subtraction of ",Num_1," &",Num_2," is", Num_1-Num_2)
elif Operand == "*":
    print("Multiplication of ",Num_1," &",Num_2," is", Num_1*Num_2)
elif Operand == "/":
    print("Division of ",Num_1," &",Num_2," is", Num_1/Num_2)
elif Operand == "**":
    print("For power of both numbers Base is ",Num_1," and Power is ",Num_2," is", Num_1**Num_2)
else:
    print("You entered wrong choice")
    
```

```python
# 2. Write a python program to check if there is any numeric value in the list using for loop.

Test_List = ["Blue", "Green", "White", 9]
Final_List = []
print("We have the list ",Test_List)
for integer in Test_List:
    if isinstance(integer, (int, float)):
        Final_List.append(integer)
print("We have the Numeric Value in the list Final_List",Final_List)
    
```

```python
# 4. Write a Python program to sum all the numeric items in a dictionary.

Num_Dict = {"A":10, "B":20, "C": 30, "D": 40, "E": 50}
Total = 0
for n in Num_Dict.values():
    Total += n
print("sum of all the numeric items in a dictionary is : ",Total)
```

```python
# 5. Write a Python program to identify duplicate values from list.

List_New = [1,2,3,4,5,3,6,8,4,9,3,0]
Final_List = []
for num in List_New:
    if List_New.count(num) > 1 and num not in Final_List:
        Final_List.append(num)
print("Duplicate numbers are :",Final_List)

    
```
