```python
# Write a python program which takes 5 inputs from the user for different subject's marks, total it and generate mark sheet using grades.

Eng = int(input("Enter MArks of English: "))
Urdu = int(input("Enter MArks of Urdu: "))
Maths = int(input("Enter MArks of Maths: "))
Computer = int(input("Enter MArks of Computer: "))
Physics = int(input("Enter MArks of Physics: "))
Total_Marks = (Eng + Urdu + Maths + Computer + Physics)
Percentage = ((Total_Marks / 500) * 100)
if Percentage <= 100 and Percentage >= 90:
    Grade = ("A1")
elif  Percentage < 90 and Percentage>= 80:
    Grade = ("A")
elif  Percentage < 80 and Percentage>= 60:
    Grade = ("B")
elif  Percentage < 60 and Percentage>= 45:
    Grade = ("C")
elif  Percentage < 45 and Percentage>= 33:
   Grade = ("D")
else:
    print("Take Exam again")

print("Your Marks For English is : ",Eng)
print("Your Marks For Urdu is : ", Urdu)
print("Your Marks For maths is : ", Maths)
print("Your Marks For Computer is : ", Computer)
print("Your Marks For Physics is : ", Physics)
print("Your Grade is : ", Grade)
print("Your percentage is :", Percentage)
```

```python
# Write a python program which accepts the radius of a circle from the user and compute the area.

import math
Radius = int(input("Enter the Radius of the circle in centimers : "))
Area = math.pi * (Radius * Radius)
print("The Area of the Circle is: ", Area,"cm")
```

```python
# Write a program which take input from the user and identify that the given number is even or odd?

First = int(input("Enter the First Number : "))
Second = int(input("Enter the Second Number : "))
Total = (First + Second)
if Total % 2 == 0:
    print("The resulting number", Total, "is an Even number")
else:
    print("The resulting number", Total, "is an Odd number")
    
```
